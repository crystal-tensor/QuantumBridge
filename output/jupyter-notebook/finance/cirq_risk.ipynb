## 2. 原理与步骤 / Model & workflow

假定损失 $L=(0,1,4,12)$，物理概率 $p=(0.55,0.25,0.15,0.05)$。
用受控 RY 制备 $\sum_j\sqrt{p_j}|j\rangle$，测量得到经验概率。
$VaR_\alpha=\inf\{l:F(l)\ge\alpha\}$；离散 CVaR 使用期望短缺定义
$ES_\alpha=VaR_\alpha+\frac{\sum_jp_j(L_j-VaR_\alpha)^+}{1-\alpha}$，
不能误用 $E[L|L\ge VaR]$。本例 $\alpha=0.85$，精确 EL=1.45、VaR=4、ES=20/3。

定义合成损失与概率 → 条件概率转 RY → 检查振幅平方 → 4096 次本地测量 → 经验 CDF → EL / VaR / ES → DKW 分位不确定区间。

## 3. 运行环境 / Local environment

需要已安装 `cirq`、`numpy`、`matplotlib` 与 Jupyter。代码不依赖本仓库私有模块，可在安装上述依赖的 Python 内核从上到下运行。没有自动安装、联网或密钥读取。

```python
import json, sys
import cirq
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML

SEED, SHOTS = 20261101, 4096
simulator = cirq.Simulator(dtype=np.complex128, seed=SEED)
plt.rcParams.update({'font.size': 10, 'axes.spines.top': False, 'axes.spines.right': False})
print('Backend: local cirq.Simulator | Cirq:', cirq.__version__)
print('Python:', sys.version.split()[0], '| hardware jobs: 0')
print('Bit order: q0 first / most significant bit; synthetic data only.')

def probabilities(circuit, qubits):
    state = simulator.simulate(circuit, qubit_order=qubits).final_state_vector
    return np.abs(state)**2

def measured_counts(circuit, qubits, key='out'):
    measured = circuit + cirq.Circuit(cirq.measure(*qubits, key=key))
    histogram = simulator.run(measured, repetitions=SHOTS).histogram(key=key)
    return np.array([histogram.get(i, 0) for i in range(2**len(qubits))], dtype=int)

def compare_plot(labels, exact, sampled, title):
    x = np.arange(len(labels))
    fig, ax = plt.subplots(figsize=(8, 3.3), layout='constrained')
    ax.bar(x-.18, exact, .36, label='Exact model', color='#137f76')
    ax.bar(x+.18, sampled, .36, label='4096 local Cirq shots', color='#bc4164')
    ax.set(xticks=x, xticklabels=labels, ylabel='Probability', title=title)
    ax.legend()
    plt.show()
```

## 4. 输入与电路 / Inputs & circuit

```python
qubits = list(cirq.LineQubit.range(2))
p, losses, alpha = np.array([.55,.25,.15,.05]), np.array([0.,1.,4.,12.]), .85
```

## 5. 中间结果与实际本地采样 / Intermediate & sampled results

图中 Exact 为同一模型的精确分布；Cirq shots 为有限次模拟测量，不是真机回执。

```python
def prepare_four(probabilities, qubits):
    p = np.asarray(probabilities, dtype=float)
    assert p.shape == (4,) and np.all(p > 0) and np.isclose(p.sum(), 1)
    left = p[0] + p[1]
    circuit = cirq.Circuit(cirq.ry(2*np.arccos(np.sqrt(left)))(qubits[0]))
    circuit.append(cirq.ry(2*np.arccos(np.sqrt(p[0]/left)))(qubits[1])
                   .controlled_by(qubits[0], control_values=[0]))
    circuit.append(cirq.ry(2*np.arccos(np.sqrt(p[2]/(1-left))))(qubits[1])
                   .controlled_by(qubits[0], control_values=[1]))
    return circuit

circuit = prepare_four(p, qubits)
exact = probabilities(circuit, qubits)
assert np.allclose(exact, p, atol=1e-12)
print('Prepared state probabilities:', exact)
print('Circuit:', circuit, sep='\n')
counts = measured_counts(circuit, qubits)
empirical = counts/SHOTS
def risk(probability):
    index = int(np.searchsorted(np.cumsum(probability), alpha))
    var = float(losses[index])
    return {'EL': float(probability@losses), 'VaR': var,
            'CVaR_ES': float(var+probability@np.maximum(losses-var, 0)/(1-alpha))}
reference, sampled = risk(p), risk(empirical)
assert np.isclose(reference['EL'], 1.45)
assert reference['VaR'] == 4 and np.isclose(reference['CVaR_ES'], 20/3)
epsilon = float(np.sqrt(np.log(2/.05)/(2*SHOTS)))
quantile = lambda a: float(losses[min(np.searchsorted(np.cumsum(empirical), a), 3)])
var_interval = [quantile(max(0, alpha-epsilon)), quantile(min(1, alpha+epsilon))]
compare_plot([str(v) for v in losses], p, empirical, 'Loss scenarios, not market observations')
plt.figure(figsize=(8,3))
plt.step(losses, np.cumsum(p), where='post', label='Exact CDF', color='#137f76')
plt.step(losses, np.cumsum(empirical), where='post', label='Sampled CDF', color='#bc4164')
plt.axhline(alpha, color='black', ls='--', label='alpha=.85')
plt.xlabel('Loss'); plt.ylabel('CDF'); plt.legend(); plt.tight_layout(); plt.show()
evidence = {'qubits': 2, 'shots': SHOTS, 'counts': counts.tolist(),
            'exact': reference, 'sampled': sampled,
            'var_95pct_DKW_interval': var_interval, 'dkw_epsilon': epsilon,
            'ideal_probability': exact.tolist(),
            'validation': 'Prepared distribution and independent exact EL/VaR/ES checked'}
```

```python
import json
from pathlib import Path
RUN_CLOUD = False
APPROVED_MAX_TASKS = 0
if RUN_CLOUD and True:
    import quafu
    from quafu_case_runtime import run_manifest
    # Run beside the extracted reproduction bundle. The default branch below
    # does not read local credentials or require any downloaded files.
    data = json.loads(Path("finance--cirq_risk.json").read_text())
    frozen = json.loads(Path("execution-plan.json").read_text())
    ids = [e["kernel_id"] for e in data["case"]["experiments"]]
    case_plan = {"kernels": {key: frozen["kernels"][key] for key in ids}}
    client = quafu.Client()  # existing SDK identity; never write a key here
    journal = run_manifest(client, case_plan, Path("finance--cirq_risk-private-journal.json"),
                           max_tasks=APPROVED_MAX_TASKS, execute=True, timeout=30)
    for entry in journal["jobs"].values():
        print(entry.get("job_id"), entry.get("state"))
else:
    print("Cloud submission disabled; historical and fresh local outputs are separate.")


import json, math
import numpy as np
data = json.loads('{"id": "finance/cirq_risk", "direction": "risk", "records": [{"name": "cirq_risk", "counts": {"10": 155, "00": 583, "01": 244, "11": 42}, "shots": 1024, "labels": ["00", "01", "10", "11"], "ideal": [0.5500000000716667, 0.24999999992942262, 0.14999999998104538, 0.05000000001786577]}], "inputs": {"probabilities": [0.55, 0.25, 0.15, 0.05], "losses": [0.0, 1.0, 4.0, 12.0], "confidence": 0.85}, "intermediate": {}, "reference": {}}')
observations, metrics = {}, {}
for row in data['records']:
    labels, shots = row['labels'], row['shots']
    p = np.array([row['counts'].get(label, 0)/shots for label in labels])
    ideal = np.asarray(row['ideal'], float)
    observations[row['name']] = (labels, p, shots)
    metrics[row['name']] = {'shots_returned': shots,
        'total_variation_from_frozen_ideal': float(.5*np.abs(p-ideal).sum()),
        'maximum_probability_difference': float(np.max(np.abs(p-ideal)))}
business = {'acceptance': 'NOT_ACCEPTED_NO_PREREGISTERED_HARDWARE_THRESHOLD',
    'scope': 'Descriptive hardware statistics, not production or quantum-advantage acceptance'}
rows = data['intermediate'].get('candidates')
if rows:
    labels, p, shots = observations['qaoa']
    by_label = {row['bitstring']:row for row in rows}
    assert set(by_label) == set(labels)
    feasible = [r for r in rows if r['feasible']]
    observed = [r for r in feasible if p[labels.index(r['bitstring'])] > 0]
    optimum = data['reference']['best_objective']
    best = min(observed, key=lambda r:r['objective']) if observed else None
    business.update(feasible_probability=float(sum(p[labels.index(r['bitstring'])] for r in feasible)),
        optimal_probability=float(sum(p[labels.index(r['bitstring'])] for r in feasible if abs(r['objective']-optimum)<1e-12)),
        expected_penalized_cost=float(sum(p[labels.index(r['bitstring'])]*r['penalized_cost'] for r in rows)),
        best_observed_feasible=best, best_observed_objective_gap=None if best is None else best['objective']-optimum,
        frozen_classical_optimum=optimum, decision_rule='best feasible actually measured, not most frequent')
def moment(name, values):
    labels, p, shots = observations[name]
    values = np.asarray(values, float)
    assert values.shape == p.shape
    mean = float(values@p)
    se = float(np.sqrt(max(0., (values-mean)**2@p)/(shots-1)))
    return {'estimate':mean, 'sampling_standard_error':se,
        'approximate_95pct_sampling_interval':[mean-1.96*se,mean+1.96*se],
        'excludes_systematic_noise_calibration_and_temporal_correlation':True}
if data['id']=='finance/03_derivatives_quafu':
    discount = math.exp(-data['inputs']['rate']*data['inputs']['maturity'])
    for label in ('call','put','delta'):
        business[label] = moment('risk_neutral_paths', discount*np.asarray(data['intermediate'][label+'_payoff']))
    business['call_price_absolute_error'] = abs(business['call']['estimate']-data['reference']['call_price'])
    business['call_price_reference'] = data['reference']['call_price']
    business['input_scope'] = 'Original synthetic risk-neutral binomial model, not observed stock data'
elif data['id']=='finance/05_correlation_quafu':
    for name in ('positive_correlation','negative_correlation'):
        labels, p, shots = observations[name]
        business[name] = moment(name, [1 if label[0]=='0' else -1 for label in labels])
        business[name]['frozen_reference'] = data['reference'][name]
        business[name]['absolute_error'] = abs(business[name]['estimate']-data['reference'][name])
        lower, upper = business[name]['approximate_95pct_sampling_interval']
        business[name]['reference_inside_sampling_only_interval'] = lower <= data['reference'][name] <= upper
    business['daily_return'] = moment('time_sampling', data['inputs']['asset_a_returns'])
    business['input_scope'] = data['inputs']['data_source']
elif data['direction']=='materials':
    labels, p, shots = observations['occupation']
    business['double_occupation'] = moment('occupation',[int(s[0])*int(s[1])+int(s[2])*int(s[3]) for s in labels])
    business['double_occupation_absolute_error'] = abs(business['double_occupation']['estimate']-data['reference']['U4_double_occupation_analytic'])
    business['scope'] = 'Two-site cluster occupation only; not a measured material band gap or ground energy'
elif data['direction']=='physics':
    labels, p, shots = observations['magnetization']
    business['magnetization_z0'] = moment('magnetization',[(-1)**int(s[0]) for s in labels])
    business['correlation_z0z2'] = moment('correlation',[(-1)**(int(s[0])+int(s[2])) for s in labels])
    business['exact_z0_reference'] = data['reference']['final_exact_z0']
elif data['direction']=='qkd':
    for name in ('z_basis_bit1','x_basis_bit1'):
        business[name+'_readout_error'] = moment(name,[1,0])
    business['mismatched_basis_probability_one'] = moment('mismatched_bases',[0,1])
    business['scope'] = 'Single-qubit basis experiments; not real QKD transport, key reconciliation or finite-key security'
elif data['direction']=='qrng':
    business['hardware_one_fraction'] = moment('biased_sampling',[0,1])
    business.update(ordered_shots_available=False, real_stream_health_or_extraction_validated=False,
        secure_entropy_or_randomness_certified=False,
        scope='Histogram only; never invent a time-ordered stream from aggregate counts')
elif data['id']=='finance/cirq_risk':
    labels, p, shots = observations['cirq_risk']
    losses = np.asarray(data['inputs']['losses'],float)
    alpha = data['inputs']['confidence']
    cumulative = np.cumsum(p)
    var = float(losses[min(int(np.searchsorted(cumulative,alpha)),len(losses)-1)])
    business.update(expected_loss=float(losses@p), confidence=alpha, VaR=var,
        CVaR_expected_shortfall=float(var+np.maximum(losses-var,0)@p/(1-alpha)),
        scope='Frozen four-state teaching loss distribution, not a bank risk system')
if data['id'] in ('finance/01_allocation_quafu','finance/04_constraints_quafu'):
    business['input_scope'] = data['inputs']['data_source']
print('HARDWARE_ANALYSIS_JSON='+json.dumps({'experiments':metrics,'business':business},allow_nan=False,sort_keys=True))
```
